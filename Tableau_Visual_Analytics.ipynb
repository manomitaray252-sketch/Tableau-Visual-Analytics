#Tableau Visual Analytics


1. Explain what Tableau is and why it is preferred over traditional spreadsheet tools for data analysis.
Also mention any two industries where Tableau is commonly used and explain one use case.

- Tableau is a powerful data visualization tool used to convert raw data into interactive dashboards, charts, and graphs. It is preferred over traditional spreadsheets like Excel because it can handle huge datasets without slowing down, updates data automatically, and makes visual analysis much faster and easier.
  Industries & Use Cases:
  Retail: Used to analyze customer buying trends and manage store inventory levels.  
   Healthcare: Used to track patient wait times and monitor hospital resource availability.

Q4. Sales Trend Analysis
Business Need:
The leadership team wants to understand how sales fluctuate over time to plan promotions and inventory.
Task:
Analyze Sales trends over time (monthly) to identify peak and slow periods


- 1.Drag Order Date to Columns and set it to continuous Month.  
2. Drag Sales to Rows to create a line chart.  
3. Peak periods are the high points on the line; slow periods are the low points.
